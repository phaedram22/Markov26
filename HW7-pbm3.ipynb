```python
#problem 3c
import numpy as np
import matplotlib.pyplot as plt

#rate function
def lam(t):
    return 0.5 * (1 + (t / 30)**2)

T = 120
lam_max = lam(T)

#simulate candidate points from HPPP(lam_max) on [0,T]
#number of candidate arrivals
N_candidates = np.random.poisson(lam_max * T)

#candidate arrival times are uniform(0,T), then sort them
candidate_times = np.sort(np.random.uniform(0, T, N_candidates))

#thinning
keep_probs = lam(candidate_times) / lam_max
uniforms = np.random.uniform(0, 1, N_candidates)
report_times = candidate_times[uniforms < keep_probs]

#total number of accepted reports
total_reports = len(report_times)

#expected value from part b
expected_reports = 380

print("simulated total reports:", total_reports)
print("expected total reports:", expected_reports)

#histogram per day
bins = np.arange(0, T + 1, 1)
plt.hist(report_times, bins=bins, edgecolor='black')
plt.xlabel("Day")
plt.ylabel("Number of reports")
plt.title("Simulated Flu Report Times (Days 0 to 120)")
plt.savefig("hw7-pbm3c.png",dpi=300, bbox_inches="tight")
plt.show()
```
