```python
#problem 3b
import numpy as np
import matplotlib.pyplot as plt

a = 0.99

#transition matrix 
P = np.array([
    [1-a, a,   0],
    [a,   0, 1-a],
    [0, 1-a,  a]
])

#transpose
A = P.T

#compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A = P^T:")
print(A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```
