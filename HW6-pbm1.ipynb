```python
#problem 1d
import numpy as np
import matplotlib.pyplot as plt

#transition matrix
P = np.array([
    [0,   1,   0,   0,   0],
    [1/3, 0,   2/3, 0,   0],
    [0,   1/2, 0,   1/2, 0],
    [0,   0,   2/3, 0,   1/3],
    [0,   0,   0,   1,   0]
], dtype=float)

#stationary distribution
pi = np.array([1/12, 1/4, 1/3, 1/4, 1/12], dtype=float)

#initial distribution
q0 = np.array([0, 0, 1, 0, 0], dtype=float)

#compute q_50 = q0 P^50
P50 = np.linalg.matrix_power(P, 50)
q50 = q0 @ P50

print("q50 =", q50)
print("pi  =", pi)
print("difference q50 - pi =", q50 - pi)

# plot
states = np.array([0, 1, 2, 3, 4])

plt.plot(states, q50, marker='o', label=r'$q_{50}(i)$')
plt.plot(states, pi, marker='s', label=r'$\pi(i)$')

plt.xticks(states)
plt.xlabel("State i")
plt.ylabel("Probability")
plt.title(r"Comparison of $q_{50}(i)$ and $\pi(i)$")
plt.legend()
plt.grid(True)
plt.savefig("stat_dist.png", dpi=300, bbox_inches="tight")
plt.show()
```

Because the chain has period 2, the distribution oscillate between even and odd states. Since the intital state is even, q50 places probability on even states. However, the stationary distribution places probability on all states. Therefore it is expected that the chain does not converge to the distribution, even for large values of n.
