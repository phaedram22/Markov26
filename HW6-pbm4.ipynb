```python
#problem 4b
import numpy as np

#parameters
a = 0.49
p_reproduce = 1 - a

#simulation settings
num_simulations = 10**3
max_generations = 200

def simulate_avalanche(a, max_generations=200, x0=1):
    #simulate one binary gw process
    #returns true if extinct before or at max_generations, else false
    
    x = x0
    
    for _ in range(max_generations):
        if x == 0:
            return True
        
        #number of particles that reproduce
        z = np.random.binomial(x, 1 - a)
        
        #each reproducing particle makes 2 offspring
        x = 2 * z
    
    return x == 0  #extinct by generation 200 or not

#run many simulations
extinct_count = 0

for _ in range(num_simulations):
    if simulate_avalanche(a, max_generations):
        extinct_count += 1

est_extinction_prob = extinct_count / num_simulations

print("est extinction probability by generation 200:", est_extinction_prob)
```

Using part a: probability of extinction = 0.49 / (1-0.49) = 0.961 approx. Because of the large number of generations, it is expected that the simulated probability would be close to the theoretical probability.
