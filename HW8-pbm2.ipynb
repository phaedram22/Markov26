```python
#part c
import sympy as sp

t = sp.symbols('t', real=True)
pF = sp.Function('pF')
pN = sp.Function('pN')
pD = sp.Function('pD')

eq1 = sp.Eq(sp.diff(pF(t), t), -2*pF(t) + pN(t) + pD(t))
eq2 = sp.Eq(sp.diff(pN(t), t), pF(t) - pN(t))
eq3 = sp.Eq(sp.diff(pD(t), t), pF(t) - pD(t))

sol = sp.dsolve(
    (eq1, eq2, eq3),
    ics={
        pF(0): 0,
        pN(0): 1,
        pD(0): 0
    }
)

for s in sol:
    print(s)
```
